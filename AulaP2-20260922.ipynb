### Exercício 8: A Média de N números
- Leia um número N;
- Leia N números;
- Imprima a média dos N números.

````python
# Solução do exercício 8

n = int(input("Introduza um número N"))
soma = 0
contador = 0
while contador < n:
    numero = int(input(f"Introduza o {contador + 1}º número: "))
    soma = soma + numero
    contador = contador + 1
if n > 0:
    media = soma / n
    print(f"A média é :{media}")
else:
    print("N tem de ser superior a 0!")
````
