### Desafio: Remover todos elementos iguais a um dado elemento

```python
# Remover todos elementos iguais a um dado elemento
def removeAll(lista, elem):
    res = []
    for i in lista:
        if i != elem:
            res.append(i)
    return res

def removeAll2(lista, elem):
    while elem in lista:
        lista.remove(elem)
    return lista

cores2 = cores2 + ["verde", "verde"]
cores2.insert(1, "verde")
cores2.insert(3, "verde")
cores2.insert(4, "verde")
print(cores2)
cores3 = removeAll(cores2, "verde")
print(cores3)
cores4 = removeAll2(cores2, "verde")
print(cores4)
```

## Tuplos

Um tuplo é um tipo estruturado composto por vários elementos que podem ser de qualquer tipo: tipo primitivo ou estruturado.

É uma sequência imutável de valores.

Exemplos:
* um par de coordenadas cartesianas (x, y): `(25, 65)`
* uma cor RGB (r, g, b): `(0, 0, 255)`
* um registo de uma base de dados (nome, identificador, curso, notas): `("Ana Dias", "A22345", "Engenharia Biomédica", [12,18,11,14])`

### São ordenados

```python
registo = ("José", 35, "Braga")
print(registo)
```

* São imutáveis, aqueles valores vão manter-se naquela ordem durante o tempo de vida do tuplo.

```python
print(registo[0])
print(registo[1])
print(registo[2])
```

### Desafio: Quem é o mais velho?

```python
p1 = ("Alfredo", 1977)
p2 = ("Helena", 1983)
p3 = ("Ana", 1953)
p4 = ("Francisco", 2003)
pessoas = [p1, p2, p3, p4]

def maisIdoso(l):
    res = l[0]
    for p in l[1:]:
        if res[1] > p[1]:
            res = p
    return res

print(maisIdoso(pessoas))  
```

### Desafio: Quem tem mais de x anos?

```python
p1 = ("Alfredo", 1977)
p2 = ("Helena", 1983)
p3 = ("Ana", 1953)
p4 = ("Francisco", 2003)
pessoas = [p1, p2, p3, p4]

def maisAnos(l, anos):

    
    return [p for p in l if 2026 - p[1] > anos]

print(maisAnos(pessoas, 67))
```

### Exercício: Resultado após uma jornada de futebol

```python
jornada21 = ('Jornada XXI', [(('Benfica',1),('Moreirense',3)), 
                             (('Sporting',2),('Braga',3)), 
                             (('Porto',3),('Rio Ave', 3))])
print(jornada21)
```

```python
def resultado(jornada):
    resultado = []
    jogos = jornada[1]
    for visitado, visitante in jogos:
        if visitado[1] < visitante[1]:
            resultado.append((visitado[0],0))
            resultado.append((visitante[0],3))
        elif visitado[1] > visitante[1]:
            resultado.append((visitado[0],3))
            resultado.append((visitante[0],0))
        else:
            resultado.append((visitado[0],1))
            resultado.append((visitante[0],1))
    return resultado

def verResultado(rlista):
    print("Equipe     Pontos")
    print("-----------------")
    for r in rlista:
        print(r[0], r[1])
    return

print(resultado(jornada21))
verResultado(resultado(jornada21))
```

### Abertura: open(nome, modo)

Modos de abertura de um ficheiro:
* "r" - Read - Valor por omissão. Abre o ficheiro para leitura, devolve error se o ficheiro não existir;
* "a" - Append - Abre o ficheiro para acrescentar, cria o ficheiro se este não existir;
* "w" - Write - Abre o ficheiro para escrever, cria o ficheiro se este não existir, apaga o conteúdo se este existir;
* "x" - Create - Cria o ficheiro, devolve error se o ficheiro já existir.

Parâmetro adicional ao modo:
* "t" - Text - Valor por omissão. Ficheiro textual;
* "b" - Binary - Ficheiro binário, por exemplo, imagens.
