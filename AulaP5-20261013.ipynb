```python
# Solução de 2.c)

def pot(a, b):              # a elevado a b
    res = 1
    while b > 0:
        res = a * res
        b = b - 1
    return res



def calcPolinomio(p, x):
    res = 0
    for termo in p:
        coe = termo[0]
        exp = termo[1]
        #coe, exp = termo
        res = coe * pot(x, exp) + res
    return res

p2 = [(2,2), (5,1), (10, 0)]
calcPolinomio(p2,1)
```

#### j) Defina uma função para guardar uma lista de polinómios num ficheiro. Utilize-a para guardar os polinómios criados até ao momento.

```python
p1 = [(1,1)]

p2 = [(2,1), (1,0)]

p3 = [(7,5), (6,3), (-2,2), (27,0)]

listap = [p1,p2,p3]
print(listap)
```

```python
# Solução de 1.j
def guardarPolinomios(listap, fnome):
    file = open(fnome,"w") # "r", "w"
    for pol in listap:
        for termo in pol:
            coe, exp = termo
            file.write(f"{coe}@{exp}|")
        file.write("\n")
    file.close()

guardarPolinomios(listap,"polinomios.txt")
```

#### k) Defina uma função para recuperar um polinómio dum ficheiro. Utilize-a para ler o ficheiro criado na alínea anterior.

```python
# Solução de 1.k)
def recuperarPolinomios(fnome):
    listap = []
    file = open(fnome,"r")
    for line in file:
        polinomio = []
        termos = line.split("|")
        for termo in termos[:-1]:
            coe, exp = termo.split("@")
            termo = (int(coe), int(exp))
            polinomio.append(termo)
        listap.append(polinomio)
    file.close()
    return listap

print(recuperarPolinomios("polinomios.txt") )
```
